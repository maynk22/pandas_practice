```python
import pandas as pd
```

```python
df = pd.read_csv('./data/results.csv')
display(df)
```

```python
#groupby: This is used to get the dataframe grouped by a certain column. It will be having aggregated values as well.

#single groupby column. 
# dropna: This parameter is used to include or exclude NaN/Nulls based on value provided to the parameter.
#reset_index: when we using aggregation function along with groupby, the aggregate function will not be having any label, so to provide that we use it along with the name of the column.
df_single_group = df.groupby('medal', dropna=False).size().reset_index(name='No_of_Medals')
display(df_single_group)

#multiple columns in groupby
df_mul_grpby = df.groupby(['noc', 'discipline'], dropna=False)['athlete_id'].nunique().reset_index(name='Num_of_Athelete')
display(df_mul_grpby)

#multiple columns in groupby along with multiple aggregations.
#agg(): We use it to perform the aggregation on the columns we want to add.
#There is two variation of agg function that we can use:
#1. When only uising single column for aggregation then, agg({'aggregate_col_name' = ('col_name', 'aggregation)})
#2. When using multiple columns in aggregation then, agg('aggregate_col_name' = ('col_name', 'aggregation'))
df_mul_agg = df.groupby(['noc', 'discipline', 'medal']).agg(
    Num_of_Athletes =('athlete_id', 'nunique'),
    Num_of_Medals =('medal', 'nunique')
)
display(df_mul_agg)

#transforms: It is used when we want to do groupby, but doesn't want to reduce the numbe rof rows. Like keeping the same df and adding a new column with the aggregation performed.
df['Avg_Country'] = df.groupby('discipline')['athlete_id'].transform('mean')
display(df)

#Normalization: It basically means scaling data to common range or value where the values become comparable.
df['Normalized'] = df.groupby('noc')['place'].transform(lambda x: x/x.mean())
display(df)
```
