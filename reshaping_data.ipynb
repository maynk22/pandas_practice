```python
import pandas as pd
```

```python
df = pd.read_csv('./data/results.csv')
display(df)
```

```python
#Pivot table:
df_pivot = pd.pivot_table(df,
                          values= 'place',
                          index=['discipline', 'event'],
                          columns=['noc', 'medal'],
                          fill_value=0)
display(df_pivot)
```

```python
#melt: It's used to convert columns into rows, basically reduce the width of the table and make it long. It's reverse of pivot.
df_melt = pd.melt(df,
    id_vars=['noc', 'discipline', 'event'],
    value_vars=['place'],
    var_name='Event',
    value_name='standing'
    )
display(df_melt)
```
