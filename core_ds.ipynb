```python
#importing pandas module
import pandas as pd
```

```python
#creating series, it can contain any type of values and index.

pd_series = pd.Series([1, 3.14, True, 'apple', [1,2,3]], index=['a', 'b', 'c', 1, [1,2]])

print(pd_series)

#print(pd_series['a'])
# this is give issue, coz list([1,2]) present in the index is unhasable as list in pandas are mutable., coz of it when pandas index for faster retrieval it throws error. 
#We will not get error when printing the entire series like we did in first print, coz python lazily build internal index mapping and this is not detected that time.
#What we can do now is, use tuple as key (1,2) as tuple are immutable hence hashable.

pd_series_tupple = pd.Series([1, 3.14, True, 'apple', [1,2,3]], index=['a', 'b', 'c', 1, (1,2)]) #this will work, coz here instead of index having list, we have tuple.
print(pd_series_tupple['a'])
```

```python
# we can create dictionary and then pass it on to the Series.

data = {'a': 1, 'b': 3.14, 'c': True, 1: 'mango', (1,2): [1,2,3]}

dict_series = pd.Series(data)

print(dict_series)

print(dict_series[(1,2)])
```
