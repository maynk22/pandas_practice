In this section we will be discussing about the Selection and Indexing in pandas.

```python
import pandas as pd
```

```python
#reading csv data using pandas

df = pd.read_csv('./data/results.csv')
print(df)
print(df.columns)
```

```python
#Selecting columns

#returing series
df_series = df['noc']
print(df_series)

#returning as dataframe. Output will contain column name.
df_df = df[['noc']]
print(df_df)
```

```python
# loc and iloc

#loc: It's label based slicing method. here we need to pass index and column labels as input and it will fetch us the data. 
#syntax: loc[row_label, column_label]

#this will return the row with label 0
df_label = df.loc[0]   
print(df_label)

#using both row and column label
df_labels = df.loc[0, ['year']]
print(df_labels)

#multiple rows and column values. It's inclusive of start and end values mentione din row label.
df_multiple = df.loc[0:2, ['year', 'noc']]
print(df_multiple)

#boolean: In this we can provide a condition and it will returns rows matching the condition
df_bool = df.loc[df['year'] > 1912, ['year','discipline']]
print(df_bool)


#iloc: It's integer position based slicing method. That means we need to pass row and column values in integer format.
#syntax: iloc[row_integer, column_integer]'

#this will return the entire row at 2nd position
df_int = df.iloc[2]
print(df_int)

#this will return the row indexed at 0,1 (excluding the end index) and columns positioned at 1, 3, 4(includes column values). iloc is exclusive of the end column, so for example if you mention in row position as [0:4], it will fetch rows 0,1,2,3 and noth 4.
df_ints = df.iloc[:2, [1,3,4]]
print(df_ints)

#this will return all rows for column positioned at 3
df_all_rows = df.iloc[:, 3]
print(df_all_rows)

#boolean: We can specify condition based indexing as well. So in the row column detail, we can mention a condition, and all the rows that pass the condition will be returned.
df_bool_iloc = df.iloc[(df['year']> 1920).values, [2,4,7]]
print(df_bool_iloc)

#important things to note.
#Consider you slice off the the row which has index label (row_label) as 3. Then in that case, if we use loc[3] it will give us error, coz it's label based and we have removed that label row. But if we use iloc, it will work coz it's index based and if 3rd index is removed, the next row becomes index 3.
df_sliced = df.iloc[1:]
df_sliced_loc = df_sliced.loc[0]
print(df_sliced_loc) #this will throw an error, since row with label 1 doesn't exist.
df_sliced_iloc = df_sliced.iloc[0]
print(df_sliced_iloc) #this will work, since it's index based not label named based
```

```python
#Boolean Indexing

#Filtering with condition: We provide filter out the dataframe based on the condition specified.

df_filtered = df[(df['year'] > 1920) & (df['noc'] == 'FRA')]
display(df_filtered)
```
