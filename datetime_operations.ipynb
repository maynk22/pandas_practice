```python
import pandas as pd
```

```python
#TimeSeries: In this section we will talk about how to perform Time series operation on Dataframe.
#date_range: It's a method in pandas library, which is used to create range of dates. Parameter: start date, end date, period:indicating how many dates to generate from starting_date, freq: determines it should be days, month, sec, mins.
#It returns DatetimeIndex
dates = pd.date_range('2026-05-01', periods=5, freq='D')
values = [100, 120, 110, 130, 125]

dates_y = pd.date_range('2026-05-01', periods=5, freq='YE')

df_ts = pd.DataFrame({'Date': dates_y, 'Value': values})
display(df_ts)

#Extracting days, years, month.

df_ts['days'] = df_ts['Date'].dt.day
df_ts['day_name'] = df_ts['Date'].dt.day_name()
df_ts['month'] = df_ts['Date'].dt.month
df_ts['year'] = df_ts['Date'].dt.year

display(df_ts)

#formatting date: We can format the date format by passing format attribute to the data.
df_date = pd.to_datetime('2026-05-18', format="%d/%m/%Y")
display(df_date)
```

```python
#Rolling Windows: It means taking a fixed sized window(num of rows) and performing calculation over that moving window.
#Scenario: If we want to find the rolling average of sales, we can use it. We can specify window like we want rolling average from third day.
df_ts['rolling_mean'] = df_ts['Value'].rolling(window=3).mean()
df_ts['rolling_std'] = df_ts['Value'].rolling(3, center=True).mean() # here center will make the output of the rolling window into centre columns of the table, irrespective of the window timeline. Run this snippet to understand what it means.

display(df_ts)
```

```python
#expanding: It is like rolling function, but without the window and it keeps of expanding. So in rolling after window it will take previous and current for any aggregation applied, but in case of expanding it will take from all previous till current while applying aggregation.
#Difference wrt rolling is rolling keep the window size fixed, so if window = 3 it will keep it fixed and take the current + previous 2, but expanding will start from min_period value as index and consider all previous values. 
df_ts['expanding_mean'] = df_ts['Value'].expanding(min_periods=1).sum() #min_period refers to from which index the expanding should start.

#emw: Exponentially Weighted Moving.
#It applies exponentially decreasing weight to older values and hgiher weight to latest value. That means in rolling it applies same weight to all values, so when we do any aggregation on that it takes full value and provides result. But in this case it assigns weights to values, so older values will not be full it will be a percent of their total value.
#formula for weight = (1-@)^t. where @(alpha) is smoothing factor and t is the age of the observation. Higher @ = more weight on the recent data.
df_ts['ewm_mean'] = df_ts['Value'].ewm(span=2).sum() #min_period: It will start the analysis after the specified value.
display(df_ts)
```
