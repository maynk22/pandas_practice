```python
import pandas as pd
```

```python
df = pd.read_csv('./data/results.csv')
display(df)
```

```python
#merge: It's used to join the tables.
#Syntax: df1.merge(df2, on="joiningg_key", how='type-of_join')

#inner join
df_merge = df.merge(df, on='year', how='inner')
display(df_merge)

#left
df_merge_left = df.merge(df, on='year', how='left')
display(df_merge_left)

#right
df_merge_left = df.merge(df, on='year', how='right')
display(df_merge_left)
```
