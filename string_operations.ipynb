```python
import pandas as pd
```

```python
df = pd.read_csv('./data/results.csv')
display(df)
```

```python
#String methods: They are vectorized methods, which are NaN safe.

df_upper = df['as'].str.upper()
display(df_upper)

df_lower = df_upper.str.lower()
display(df_lower)

df_title = df_lower.str.title()
display(df_title)

df_strip = df_title.str.strip()
display(df_strip)

df_contains = df_title.str.contains("ean") #returns true for matching values, false which don't match.
display(df_contains)

df_split = df_title.str.split(" ", expand=True) #split the string based on delimeter provided in start, expand true means create a new column for each split value.
display(df_split)
```
