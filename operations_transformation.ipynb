```python
import pandas as pd
```

```python
df = pd.read_csv('./data/results.csv')
display(df)
```

```python
#Adding columns

#map: It's used to transform the data. Map is applied to series or single column values.
df['has medal'] = df['medal'].notna().map({True: 'yes', False: 'no'}) #adding a new column based on other column value.
display(df)

#it can be applied to dictionary.
standing = {
    1: '1st place',
    2: '2nd place',
    3: '3rd place'
}

df['Standing'] = df['place'].map(standing)
display(df)

#apply: This is also used to transform data, but it can be applied to series, dataframe rows and columns.
df['is_medal'] = df['medal'].apply(lambda x: 1 if x == True else 0)
display(df)
```

```python
#Sorting and Ranking

#sort_values: This function is used to sort the data, based on the column provides, multiple columns can also be provides in a list. We can change order based on ascending or decending keeping ascending flag true or false. Default is ascending true.
df_sorted = df.sort_values('year', ascending=False)
display(df_sorted)

df_sorted_multiple = df.sort_values(['year', 'type'], ascending=[True, False])
display(df_sorted_multiple)

#rank: It's to provide ranking to the data using specific column.
#parameters: 
# 1. ascending: Used to rank lowest to highest or highest to lowest. Default is lowest to highest.
# 2. method: This determines what will happen if two values are same. There are in total 5 methods:
#   i. average: It's default. So if there's a tie, it will take the average of the tie ranks and provide that value as rank value.
#   ii. min: It will provide both the values minimum rank. Like if for rank =3 there are three same values, it will assign 3 to all. Then it will skip the next n ranks (here n is number of tie values) and then provide next rank after the gap.
#   iii. max: It will provide both the values maximum rank. Like if for rank =3 there are three same values, it will assign 5 to all. Then it will skip the next n ranks (here n is number of tie values) and then provide next rank after the gap.
#   iv. first: It will provide the first occurence of tie value as lower rank and then subsequent values, subsequent ranks. Here no rank are skipped.
#   v. dense: here it will provide same rank to same value occurences, but it will not skip ranks like min and max.
# 3. pct: It is used to provide the rank in percentage. Range 0.1 - 1.0.
df_rank = df['place'].rank()
display(df_rank)
```

```python
#filter: It's used to flter dataframe based on the label names. Column and row label names. Not filtering the df based on conditions applied to values in df.
#For example: We will filter the df based on lebles for columns and rows. By default it applies filter on column labels, but by changing axis = 0, we can make it apply on rows.
df_filtered = df.filter(items=['year', 'type'])
display(df_filtered)

df_filtered_rows = df.filter(items=[0,3,6,8], axis=0)
display(df_filtered_rows)

#like: It will filter the df with the columns, that meets the condition.
df_filter_like = df.filter(like='dis')
display(df_filter_like)

df_filter_grpby = df.groupby('noc').filter(lambda x: x['place'] <3)
display(df_filter_grpby)
```
