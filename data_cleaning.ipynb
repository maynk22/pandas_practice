```python
import pandas as pd
```

```python
df = pd.read_csv('./data/results.csv')
display(df)
```

```python
#Detecting and Handling NaN values

#isnull() : It checks against each column and row, and returns a df showing true if its null and false if its not. Alternate to it is using .isna() -> It's like alias to isnull().
df_nulls = df.isnull()
display(df_nulls)
display(df_nulls.sum())  # It will show the count of nulls/NaN per column.

#notnull(): It checks against each column and row, and returns a df showing true if its not null and false if its not.
df_notnull = df.notnull()
display("/notnull", df_notnull)

#fillna(): It is used to replace null/NaN values in the df with the specified value.
df_fillna_full = df.fillna(0)
display(df_fillna_full)

df_fillna_column = df[['year']].fillna(0)  #it will only replace nulls/NaN with 0 for this specific column. Returns only the row where it's applied.
display(df_fillna_column)

df_fillna_mulcol = df.fillna({'noc': 'unknown', 'team': 'No Team'}) #it's used to fill multiple columns null/NaN with values.
display(df_fillna_mulcol)

df_ffill = df[['noc']].ffill(limit=3) #ffill method: It's a forward fill method, in this it's previous not null value present in the row and fill next rows with those values. We can provide limit, which means apply to next n consecutive null values. So it will fill next n consecutive nulls with the not null value.
display(df_ffill)

df_bfill = df[['team']].bfill() #bfill: It's backfill method, here it takes next null values present in the rows and fill the previous null rows.
display(df_bfill)

#dropna: It will drop the rows which has null/NaN values.
df_dropna = df.dropna()
display(df_dropna)

df_dropna_col = df[['year']].dropna()  #this will return only that row on which dropna is applied.
display(df_dropna_col)

df_dropna_axis = df.dropna(axis=1) #here axis refer to either row or column drop. So axis=1 drop the column with null vlaues, axis=0 will drop the row with null values.
display(df_dropna_axis)

df_dropna_how = df.dropna(how='all') #here how is a conditional drop. So if how='all' then it will drop the column only if all values are null, if how='any' then it will drop the column if there are any null values.
display(df_dropna_how)

df_dropna_thres = df.dropna(thresh=3) #thresh, refers to threshold value of not null for the rows need to be kept. Here so in the mentioned example the row will be kept if it contains 3 or more non null values.
display(df_dropna_thres)
```

```python
#Removing Duplicates

#duplicated(): It returns each index mentioning True or False based on if it's a duplicated row or not.
df_dup = df.duplicated()
print(df_dup)

df_dup_sum = df_dup.sum()  #it returns the count of total duplicate rows in the dataframe.
print(df_dup_sum)

df_subset = df.duplicated(subset=['noc']) #here subset is used to check the duplicates in that particular column.
display(df_subset)

df_keep = df.duplicated(keep='last') #here keep refes to which value to keep. By default the first occurence of the value is considered unique and rest other occurences as duplicates. Keep has two values, first and last. First means keep first value occurence as unique its default behaviour as well, and last which means keep last occurence as unique
display(df_keep)

#drop_duplicates(): It drop duplicate rows from the dataframe and returns the updated dataframe.
df_drop_dup = df.drop_duplicates()
display(df_drop_dup)
```

```python
#Type casting and renaming

#astype: It's used to convert the datatype of column.
df['Position'] = df['place'].astype('Int64') 
display(df)

#category: It's a special datatype in pandas. It's used for values which has categorical value like Male/Female, list of cities, countires, etc. This operation saves memory. Performing operations on this becomes faster and efficient.
df['medal'] = df['medal'].astype('category') # now the datatype of medal column will change from str to category.
display(df)

df['year'] = pd.to_numeric(df['year'], errors='coerce') #this is hard changing of column dtype, basically if values can change, they'll convert into numeric, those values which can't be converted into numeric will change to NaN as mentioned by errors attribute.
display(df)

#renaming
df_renamed = df.rename(columns = {'noc': 'name of country'}) #rename: can rename multiple columns at once. Format is columns={'old_name': 'new_name'}
display(df_renamed)
```
