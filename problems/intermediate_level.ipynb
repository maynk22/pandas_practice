```python
#Ques 7: Create orders and customers DataFrames, merge them, and find total spending per customer.

orders = pd.DataFrame({ "order_id": [101,102,103,104,105], "customer_id": [1,2,1,3,2], "amount": [250,400,150,300,500] }) 
customers = pd.DataFrame({ "customer_id": [1,2,3], "name": ["Alice","Bob","Charlie"], "city": ["NYC","LA","Chicago"] })

df_merged = orders.merge(customers, how='left', on='customer_id')
df_output = df_merged.groupby('customer_id')['amount'].sum().reset_index(name='total_spend')
display(df_output)
```

```python
#Ques 8: Create a pivot table of total sales by product (rows) and month (columns), including totals.

dates = pd.date_range("2024-01-01", periods=60, freq="D") 
products = np.random.choice(["A","B","C"], 60) 
sales = np.random.randint(100, 500, 60) 
df_s = pd.DataFrame({"date":dates,"product":products,"sales":sales})

df_s['month'] = df_s['date'].dt.to_period('M')

df_pivot = pd.pivot_table(df_s,
                          values='sales',
                          index='product',
                          columns='month',
                          aggfunc=sum,
                          margins=True
                          )

display(df_pivot)
```

```python
#Ques 9: Extract the email domain, count domain occurrences, and identify Gmail users.

df_e = pd.DataFrame({ "user_id": [1,2,3,4,5,6], "email": ["alice@gmail.com","bob@yahoo.com","charlie@gmail.com", "diana@outlook.com","eve@gmail.com","frank@yahoo.com"] })
df_e['domain'] = df_e['email'].str.split('@').str[1]
df_output = df_e['domain'].str.contains('gmail.com')
display(df_output)
df_count = df_e[df_e['domain'] == 'gmail.com'].count()
display(df_count)
```

```python
#ques 10: Generate 90 days of daily sales. Compute a 7-day rolling average, find the best sales week, and get monthly totals.

dates = pd.date_range("2024-01-01", periods=90, freq="D") 
sales = np.random.randint(500, 2000, 90) 
df_ts = pd.DataFrame({"date":dates,"sales":sales})

df_ts['sales_avg'] = df_ts['sales'].rolling(window=7).mean()
display(df_ts)

df_best_sales = df_ts['sales_avg'].nlargest(1)
display(df_best_sales)

df_ts['month'] = df_ts['date'].dt.to_period('M')
display(df_ts)

df_ts_monthly = df_ts.groupby('month')['sales'].sum().reset_index(name='Total_Monthly_sales')
display(df_ts_monthly)

df_best_check = df_ts.loc[15]
display(df_best_check)
```

```python
dates = pd.date_range("2024-01-01", periods=90, freq="D") 
sales = np.random.randint(500, 2000, 90) 
df_ts = pd.DataFrame({"date":dates,"sales":sales}) 
df_ts.set_index("date", inplace=True) 
df_ts["rolling_7d"] = df_ts["sales"].rolling(7).mean() 
df_ts["week_total"] = df_ts["sales"].rolling(7).sum()
best_end = df_ts["week_total"].idxmax() 
print(best_end)
best_start = best_end - pd.Timedelta(days=6) 
print(f"Best week: {best_start.date()} to {best_end.date()}") 
print(df_ts.resample("ME")["sales"].sum())
```
